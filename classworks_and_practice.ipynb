```python
class BankAccount:
    """Superclass to manage basic banking operations."""

    def __init__(self, owner, initial_balance=0.0):
        self.owner = owner
        self._balance = float(initial_balance)

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Deposited: {amount:.2f}. New Balance: {self._balance:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self._balance:
            self._balance -= amount
            print(f"Withdrew: {amount:.2f}. New Balance: {self._balance:.2f}")
        else:
            print("Insufficient funds or invalid withdrawal amount.")

    def get_balance(self):
        return self._balance


class SavingsAccount(BankAccount):
    """Subclass that inherits BankAccount and adds interest calculation."""

    def __init__(self, owner, initial_balance=0.0, interest_rate=0.04):
        # Initialize the superclass attributes
        super().__init__(owner, initial_balance)
        self.interest_rate = float(interest_rate) # e.g., 0.04 for 4%

    def calculate_interest(self, years):
        """Calculates compound interest using the formula A = P(1 + r)^t"""
        principal = self.get_balance()
        # Formula: A = P(1 + r)^t
        total_amount = principal * ((1 + self.interest_rate) ** years)
        interest_earned = total_amount - principal

        return interest_earned, total_amount

    def apply_interest(self, years):
        """Calculates and automatically deposits the interest earned."""
        interest, new_balance = self.calculate_interest(years)
        print(f"\n--- Interest Calculation for {years} Years ---")
        print(f"Interest Rate: {self.interest_rate * 100}%")
        print(f"Interest Earned: ${interest:.2f}")
        self.deposit(interest) # Uses the superclass method to update balance

# --- Demonstration ---

# 1. Create a savings account for Alice with $1000 at 5% annual interest
alice_account = SavingsAccount(owner="Alice", initial_balance=1000.0, interest_rate=0.05)

# 2. Check current balance
print(f"Initial Balance: ${alice_account.get_balance():.2f}")

# 3. Add regular transaction
alice_account.deposit(250.0)
alice_account.withdraw(100.0)

# 4. Apply compound interest over 3 years
alice_account.apply_interest(years=3)
```
