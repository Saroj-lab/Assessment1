```python

df_cleaned_rows = df.drop_duplicates()


#df = df.T.drop_duplicates().T


df_cleaned_rows
```

```python
df_clean = df_cleaned_rows.T.drop_duplicates().T
df_clean
```

```python
df_clean.isnull().sum()
```

```python
num_cols_median = [
    'Sale Price', 'No of Bathrooms', 'Flat Area (in Sqft)', 'Lot Area (in Sqft)',
    'Area of the House from Basement (in Sqft)', 'Living Area after Renovation (in Sqft)'
]
for col in num_cols_median:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

geo_cols = ['Zipcode', 'Latitude', 'Longitude']
for col in geo_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

df_clean['No of Times Visited'] = df_clean['No of Times Visited'].fillna(df_clean['No of Times Visited'].mode()[0])

print("Remaining missing values:")
print(df_clean.isnull().sum())


df_clean
```

```python
from sklearn.preprocessing import StandardScaler

df_scaled = df_clean.copy()

exclude_cols = ['Sale Price', 'ID', 'Date House was Sold', 'Waterfront View', 'No of Times Visited', 'Condition of the House']
num_cols = [col for col in df_scaled.columns if col not in exclude_cols]

scaler = StandardScaler()

df_scaled[num_cols] = scaler.fit_transform(df_scaled[num_cols].astype(float))

df_scaled
```

```python
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder
```

```python
df_clean.select_dtypes(include=['object', 'category']).columns.tolist()
```

```python
condition_map = {'Bad': 0, 'Fair': 1, 'Good': 2, 'Excellent': 3}
visit_map = {'None': 0, 'Once': 1, 'Twice': 2, 'Thrice': 3, 'Four': 4}
label_enc = LabelEncoder()
df_clean['Condition of the House'] = label_enc.fit_transform(df_clean['Condition of the House'])
df_clean['No of Times Visited'] = label_enc.fit_transform(df_clean['No of Times Visited'])
df_clean
```

```python
from sklearn.preprocessing import OneHotEncoder

onehot_enc = OneHotEncoder(drop='first', sparse_output=False, dtype=int)

encoded_array = onehot_enc.fit_transform(df_clean[['Waterfront View']])

encoded_col_name = onehot_enc.get_feature_names_out(['Waterfront View'])
df_clean[encoded_col_name] = encoded_array

df_clean.drop(columns=['Waterfront View'], inplace=True)

df_clean
```

```python
outlier_features = [
    'Sale Price',
    'Flat Area (in Sqft)',
    'Lot Area (in Sqft)',
    'Area of the House from Basement (in Sqft)'
]

for col in outlier_features:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers_count = ((df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)).sum()
    print(f"Feature '{col}': Removing {outliers_count} outlier rows.")

    df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]

df_clean
```

```python
from sklearn.model_selection import train_test_split


X = df_clean.drop(columns=['Sale Price'])
y = df_clean['Sale Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

```
